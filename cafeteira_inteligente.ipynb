## 📦 Arquitetura do Controle de Estoque

Atualmente, o controle é baseado na quantidade de **cápsulas físicas** disponíveis para cada tipo de bebida. A lógica foi desenhada em dois momentos cruciais para garantir a integridade dos dados:

1. **Momento 1 (Validação de Sessão):** O sistema calcula o "Estoque Disponível" subtraindo as cápsulas do estoque real pelas cápsulas que já estão reservadas no carrinho do usuário. Isso impede que um cliente adicione mais itens do que o totem pode entregar.
2. **Momento 2 (Commit/Baixa Real):** O estoque real só sofre a dedução física após a confirmação e fechamento da conta (`Deseja fechar a conta? (s/n)`). Se o usuário cancelar a operação, o estoque físico permanece intacto.

#### As funções de estruturação ```while True, break e pass```

```while True``` Cria um laço estruruta de reptição(infinita) onde um ```if, elif e else``` normalmente fica até que atenda uma condição e junto a essa condição o ```break``` é o interruptor de emergência, ele é a unica coisa capaz de quebrar um  ```True```  e encerrar a repetição, o ```pass``` (literalmente: "passar") ou seja caso alguma estutura de condição não esteja pronta ele simplesmente passa por ela sem interromper o fluxo de execução, de forma simples ela não faz nada, é uma instrução nula.

Exemplo:
```python
while True:
    resposta = input("Digite 'sair' para parar: ")
    if resposta == 'sair':
        break # Interrompe o loop na hora

    elif  
        (condição não esta pronta então o pass faz ignorar)
        pass
    print("Você não digitou sair, o loop continua...")```

````python
# resistore
cores_valores = {
    'preto': 0,
    'marrom': 1,
    'vermelho': 2,
    'laranja': 3,
    'amarelo': 4,
    'verde': 5,
    'azul': 6,
    'violeta': 7,
    'cinza': 8,
    'branco': 9}

multiplicado_resit = {
    'preto': 1.0,         # Multiplica por 1 (10^0)
    'marrom': 10.0,       # Multiplica por 10 (10^1)
    'vermelho': 100.0,    # Multiplica por 100 (10^2)
    'laranja': 1000.0,    # Multiplica por 1.000 (10^3)
    'amarelo': 10000.0,   # Multiplica por 10.000 (10^4)
    'verde': 100000.0,    # Multiplica por 100.000 (10^5)
    'azul': 1000000.0,    # Multiplica por 1.000.000 (10^6)
    'violeta': 10000000.0,  # Multiplica por 10.000.000 (10^7)
    'ouro': 0.1,          # Multiplica por 0.1
    'prata': 0.01         # Multiplica por 0.01
}


def calcular_resistor_3_faixas(faixa1, faixa2, faixa3):
    digito1 = cores_valores[faixa1]
    digito2 = cores_valores[faixa2]

    valor_base = (digito1 * 10) + digito2

    multiplicador = multiplicado_resit[faixa3]

    resultado = valor_base * multiplicador

    return resultado


while True:
    print(f"""
Escolha a cores do resistores cada cor vai representar uma 
resistenca mais alta ou baixa quanto maior o numero mais alta a resistencia e quando mais baixo menor :""")

    for cor, numero in cores_valores.items():
        print(f"Cor: {cor}, resistencia: {numero}")

    cores_escolhidas = []

    for i in range(1, 4):
            cor = input(f"Digite a cor da {i} faixa:").lower()

            cores_escolhidas.append(cor)

    print(
            f"\nVocê escolheu as cores: {cores_escolhidas[0]},{cores_escolhidas[1]}n e a resistencia {cores_escolhidas[2]} ")
    resultado_final = calcular_resistor_3_faixas(cores_escolhidas[0], cores_escolhidas[1], cores_escolhidas[2])    
    print(f" Seu resistor é de : {resultado_final}")

    continuar = input("\nDeseja calcular outro resistor? (s/n): ").lower()
    if continuar == 'n':
        print("Encerrando o programa... Até logo!")  
        break
````

### 📖 Glossário de Comandos Utilizados/ Revisamento

* **`def`**: Bloco de comando usado para criar funções personalizadas. Ele serve para isolar e organizar trechos de código complexos, facilitando a reutilização da lógica.
* **`dict` (Dicionário)**: Estrutura de dados que armazena informações no formato de "chave e valor" (como a tabela de cores e seus números), permitindo buscar dados de forma rápida.
* **`json`**: Formato de texto leve usado para trocar e salvar dados. No Python, ele transforma dicionários e listas em arquivos físicos que ficam guardados no computador.
* **`json.dump()`**: Comando que pega os dados do Python (listas ou dicionários) e os grava ("despeja") para dentro de um arquivo JSON físico.
* **`json.load()`**: Comando que lê o conteúdo de um arquivo JSON físico e o carrega de volta para dentro do Python em formato de lista ou dicionário.
* **`with open()`**: Gerenciador de segurança que abre um arquivo no computador e garante que ele seja fechado automaticamente quando o bloco de código termina, evitando que o arquivo corrompa.
* **`w` (Write)**: Modo de abertura de arquivo focado em **escrita**. Ele cria um arquivo novo ou limpa o conteúdo de um arquivo existente para gravar dados do zero.
* **`r` (Read)**: Modo de abertura de arquivo focado apenas em **leitura**. Ele serve para abrir um arquivo existente e ler as informações de dentro dele.
* **`encoding="utf-8"`**: Tradutor universal de caracteres que garante que o Python salve e leia o arquivo de texto com acentos e cedilhas corretamente, sem gerar símbolos estranhos.
* **`while True`**: Laço de repetição infinito que mantém o programa rodando continuamente até encontrar um comando de parada (`break`).
* **`for ... in`**: Estrutura de repetição usada para percorrer e ler, item por item, os dados de uma lista ou dicionário de forma automática.
* **`.lower()`**: Função que converte qualquer texto para letras minúsculas, padronizando a entrada de dados do usuário para evitar erros de leitura.

````python
import json


lista_filmes = []
lista_jogos = []

def salvar_no_arquivo(nome_arquivo, dados_para_salvar):
    with open(nome_arquivo, "w", encoding="utf-8" ) as arquivo:
        json.dump(dados_para_salvar, arquivo, indent=4, ensure_ascii= False)

def ler_arquivo(nome_arquivo):
    with open(nome_arquivo, "r", encoding="utf-8" ) as arquivo:
        return json.load(arquivo)

lista_filmes = ler_arquivo("filmes.json")
lista_jogos = ler_arquivo("jogos.json")

while True:
    print("\n--- MEU BLOCO DE NOTAS ---")
    opcao = input("Deseja: [1] Adicionar Nota | [2] Ver Notas | [3] Sair: ")
    
    if opcao == "3":
        break
        
    elif opcao == "1":
        categoria = input("Qual categoria? [f] Filmes ou [j] Jogos: ").lower()
        nota = input("Digite a sua anotação: ")
        
        if categoria == "f":
            lista_filmes.append(nota)
            # Agora você chama o def passando o arquivo correto e a lista atualizada:
            salvar_no_arquivo("filmes.json", lista_filmes)
            print(f"Filme {nota} salvo com sucesso!")
            
        elif categoria == "j":
            lista_jogos.append(nota)
            salvar_no_arquivo("jogos.json", lista_jogos)
            print(f"Jogo: {nota} salvo com sucesso!")
            
            
    elif opcao == "2":
        categoria = input("Qual nota deseja ver [f] Filmes ou [j] Jogos: ").lower()
        
        if categoria == "f":
            # Chamamos a ferramenta de leitura para carregar o histórico real do HD
            filmes_salvos = ler_arquivo("filmes.json")
            print(f"\nSeus Filmes: {filmes_salvos}")
        
        elif categoria == "j":
            jogos_salvos = ler_arquivo("jogos.json")
            print(f"\nSeus Jogos: {jogos_salvos}")
````

Desafio 1: O Contador de Acessos

Objetivo: Praticar a persistência de um dado simples (um número) usando JSON e funções.

Cenário: Toda vez que um usuário abre um sistema administrativo, o programa deve ler um arquivo chamado acessos.json, somar +1 ao total de acessos e salvar o novo valor.

````python
import json


def salvar_acesso(lista_dados):
    # Mudado de 'lista_acessos' para o nome do arquivo em texto: "acessos.json"
    with open("acessos.json", "w", encoding="utf-8") as arquivo:
        json.dump(lista_dados, arquivo, indent=4, ensure_ascii=False)

def ler_acesso():
    # Colocado try/except para o código não quebrar se o arquivo ainda não existir
    try:
        # Mudado de 'lista_acessos' para o nome do arquivo em texto: "acessos.json"
        with open("acessos.json", "r", encoding="utf-8") as arquivo:
            return json.load(arquivo)
    except FileNotFoundError:
        return [] # Se o arquivo não existir, começa com uma lista vazia

def atualizar_acessos():
    acessos_salvos = ler_acesso()

    while True:
        print("\nAcesso ao sistema [S] continuar [N] para sair")
        resposta = input("Continuar...? ").lower().strip()

        # Criamos o texto do novo acesso aqui
        novo_acesso = f"Acesso número {len(acessos_salvos) + 1}"

        if resposta == "n":
            print("Saindo do sistema...")
            break

        if resposta == "s":
            print("Acessando sistema...")
            acessos_salvos.append(novo_acesso)
            salvar_acesso(acessos_salvos)
            print(f"Acessos feitos até agora: {acessos_salvos}")

# Executa o código
atualizar_acessos()
````

# 📓 Notas de Estudo: Funções, Listas e Tratamento de Erros em Python

## 🛠️ Palavra-chave `def` (Funções)
* **O que é:** Bloco de comando isolado para executar uma **responsabilidade única** (evita repetição de código).
* **Como funciona:** Recebe os dados nos parênteses, executa as condições internas e retorna um resultado. Tudo o que estiver recuado (Tab) pertence ao bloco.

## 📦 Estrutura de Dados: Lista de Tuplas
* **O que é:** Uma lista `[ ]` onde cada item dentro dela é uma tupla `( )` contendo um par de informações grudadas (ex: Nome e Preço).
* **Sintaxe utilizada:** `carrinho = [("Espresso", 3.0), ("Latte", 4.5)]`
* **Leitura com `for`:** Usa duas variáveis para desembrulhar o par a cada volta: `for cafe, preco in carrinho:`

## 🔄 Operadores de Condição (`if` + `and`)
* **O que é:** Usado para misturar e exigir mais de uma validação obrigatória ao mesmo tempo.
* **Sintaxe utilizada:** `if cafe == nome_item and preco == preco_item:`
* **Regra:** O Python só executa o que está dentro do bloco se as duas condições forem verdadeiras juntas.

## 🛡️ Estrutura `try / except` (Tratamento de Erros)
* **O que é:** Um escudo para o sistema. Captura falhas previsíveis que fogem do controle do código (como inputs errados ou itens ausentes) e impede o programa de travar.
* **O bloco `try`:** Abre a chave da tentativa. Você coloca a ação perigosa que corre o risco de dar erro (ex: `carrinho.remove()`).
* **O bloco `except`:** Abre a chave do tratamento. É o plano B do sistema. Se o `try` falhar e disparar um erro específico (como `ValueError`), o `except` captura e executa uma mensagem amigável na tela.

## 🌍 Escopo Global (`global`)
* **O que é:** Comando que avisa o Python que a função vai modificar uma variável criada do lado de fora do bloco.
* **Sintaxe utilizada:** `global total_conta` (deve ser declarado logo na primeira linha interna da função).


> **Nota do Desenvolvedor:** *O avançado nada mais é do que o básico muito bem estruturado. Minha maior meta agora é persistir nos micro-desafios de prática para que a velocidade dos meus dedos alcance o ritmo da minha lógica de arquitetura.*


````python



carrinho = [("Espresso", 3.0), ("Latte", 4.5), ("Espresso", 3.0), ("Mocha", 5.5)]

total_conta = 16 # Soma de: Espresso (3.0) + Latte (4.5) + Espresso (3.0) + Mocha (5.5)

def remover_item_do_carrinho(nome_item, preco_item):
    # Usamos a palavra 'global' para o Python saber que vamos mexer na variável que está fora do def
    global total_conta
    item_para_remover = (nome_item, preco_item)
    # Outra forma que pensei sem usar try/except
#        for cafe, preco in carrinho:
#            if cafe ==  nome_item and preco == preco_item:
#                carrinho.remove((cafe, preco))
#                total_conta -= preco
#                break
    try:
        carrinho.remove(item_para_remover)
        total_conta -= preco_item
        print(f"{nome_item} removido com sucesso!")
                
    except ValueError:
        print(f"Erro: o item '{nome_item}' com o preco R$ {preco_item} não foi encontrado")


# --- TESTANDO O MECANISMO ---
print(f"Antes: {carrinho} | Total: R$ {total_conta:.2f}")

# Teste 1: Removendo um Espresso que existe no carrinho
remover_item_do_carrinho("Espresso", 3.0)
print(f"Depois do Teste 1: {carrinho} | Total: R$ {total_conta:.2f}")

# Teste 2: Tentando remover um café que não está no carrinho (deve exibir o seu aviso)
remover_item_do_carrinho("Cappuccino", 5.0)
````
