## 📦 Arquitetura do Controle de Estoque

Atualmente, o controle é baseado na quantidade de **cápsulas físicas** disponíveis para cada tipo de bebida. A lógica foi desenhada em dois momentos cruciais para garantir a integridade dos dados:

1. **Momento 1 (Validação de Sessão):** O sistema calcula o "Estoque Disponível" subtraindo as cápsulas do estoque real pelas cápsulas que já estão reservadas no carrinho do usuário. Isso impede que um cliente adicione mais itens do que o totem pode entregar.
2. **Momento 2 (Commit/Baixa Real):** O estoque real só sofre a dedução física após a confirmação e fechamento da conta (`Deseja fechar a conta? (s/n)`). Se o usuário cancelar a operação, o estoque físico permanece intacto.

#### As funções de estruturação ```while True, break e pass```

```while True``` Cria um laço estruruta de reptição(infinita) onde um ```if, elif e else``` normalmente fica até que atenda uma condição e junto a essa condição o ```break``` é o interruptor de emergência, ele é a unica coisa capaz de quebrar um  ```True```  e encerrar a repetição, o ```pass``` (literalmente: "passar") ou seja caso alguma estutura de condição não esteja pronta ele simplesmente passa por ela sem interromper o fluxo de execução, de forma simples ela não faz nada, é uma instrução nula.

Exemplo:
```python
while True:
    resposta = input("Digite 'sair' para parar: ")
    if resposta == 'sair':
        break # Interrompe o loop na hora

    elif  
        (condição não esta pronta então o pass faz ignorar)
        pass
    print("Você não digitou sair, o loop continua...")```

````python
#v.0.00

# Inicialização das variáveis do sistema
carrinho = []
total_conta = 0.0

print("--- BEM-VINDO AO TOTEM DA CAFETERIA ---")

# Este loop mantém o totem ligado
while True:
    print("\n--- MENU PRINCIPAL ---")
    print("1. Ver Menu e Comprar")
    print("2. Ver Carrinho e Fechar Conta")
    print("3. Cancelar e Sair")
    
    opcao = input("Escolha uma opção: ")
    
    if opcao == "1":
        print("\n--- NOSSOS CAFÉS ---")
        print("1. Espresso - R$ 5.00")
        print("2. Latte - R$ 7.50")
        print("3. Cappuccino - R$ 9.00")
        
        escolha_cafe = input("Digite o número do café desejado: ")
        
        # Aqui você precisa adicionar o café ao carrinho e somar o valor
        if escolha_cafe == "1":
            carrinho.append("Espresso")
            total_conta += 5.00
            print("Espresso adicionado!")
        # 1. SEU CÓDIGO AQUI: Faça o mesmo para o Latte (2) e Cappuccino (3)
        elif escolha_cafe == "2":
            carrinho.append("Latte")
            total_conta += 7.50
            print("Latte adicionado!")
        elif escolha_cafe == "3":
            carrinho.append("Cappuccino")
            total_conta += 9.00
            print("Cappuccino adicionado!")
        
        else:
            print("Opção de café inválida!")
            
    elif opcao == "2":
        print("\n--- SEU CARRINHO ---")
        print(f"Itens: {carrinho}")
        print(f"Total atual: R$ {total_conta:.2f}")
        resposta = input ("Deseja fechar a conta? (s/n) Ou seja continuar a compra")

        # 2. SEU CÓDIGO AQUI: O usuário quer fechar a conta. O que o loop deve fazer?
        if resposta == "s":
            print("Conta fechada. Obrigado por comprar!")
            carrinho = []
            total_conta = 0.0
            break

        elif resposta == "n":
            print("Continuando a compra...")

    elif opcao == "3":
        print("Pedido cancelado. Volte sempre!")
        # 3. SEU CÓDIGO AQUI: O usuário quer sair sem comprar nada. O que fazemos com o loop?
        break

    else:
        print("Opção inválida! Escolha 1, 2 ou 3.")
````

````python
#v.0.02 - Implementação de controle de estoque


# 1. Novo dicionário de controle de estoque
estoque = {
    "Espresso": 3,
    "Latte": 2,
    "Cappuccino": 1
}
carrinho = []
total_conta = 0.0

print("--- BEM-VINDO AO TOTEM DA CAFETERIA ---")

while True:
    print("\n--- MENU PRINCIPAL ---")
    print("1. Ver Menu e Comprar")
    print("2. Ver Carrinho e Fechar Conta")
    print("3. Cancelar e Sair")

    opcao = input("Escolha uma opção: ").strip()

    if opcao == "1":
        print("\n--- NOSSOS CAFÉS ---")
        
        #=== atualizar de menu com estoque temporario (tentativa) ===
        espresso_disp = estoque["Espresso"] - carrinho.count("Espresso")
        latte_disp = estoque["Latte"] - carrinho.count("Latte")
        cappuccino_disp = estoque["Cappuccino"] - carrinho.count("Cappuccino")

        # exebi a quantidade atual no menu para o cliente ver
        print(f"1. Espresso - R$ 5.00 ({espresso_disp} unid. restantes)")
        print(f"2. Latte - R$ 7.50 ({latte_disp} unid. restantes)")
        print(f"3. Cappuccino - R$ 9.00 ({cappuccino_disp} unid. restantes)")

        escolha_cafe = input("Digite o número do café desejado: ").strip()

        if escolha_cafe == "1":
            if espresso_disp > 0:
                carrinho.append("Espresso")
                total_conta += 5.00
                print("☕ Espresso adicionado ao carrinho!")
            else:
                print("❌ Desculpe, o Espresso está esgotado!")

        elif escolha_cafe == "2":
            if latte_disp > 0:
                carrinho.append("Latte")
                total_conta += 7.50
                print("🥛 Latte adicionado ao carrinho!")
            else:
                print("❌ Desculpe, o Latte está esgotado!")

        elif escolha_cafe == "3":
            if cappuccino_disp > 0:
                carrinho.append("Cappuccino")
                total_conta += 9.00
                print("🍫 Cappuccino adicionado ao carrinho!")
            else:
                print("❌ Desculpe, o Cappuccino está esgotado!")
        else:
            print("❌ Opção de café inválida!")

    elif opcao == "2":
        print("\n--- SEU CARRINHO ---")
        if not carrinho:
            print("Seu carrinho está vazio!")
        else:
            
            for item in set(carrinho):
                print(f"{item}: {carrinho.count(item)} unid.")

            print(f"Total atual: R$ {total_conta:.2f}")


            #=== logica para exibir o estoque temporario (tentativa)===
            estoque_temporario = estoque.copy()
            for item in set(carrinho):
                if item in estoque_temporario:
                    estoque_temporario[item] -= carrinho.count(item)  # Subtrai a quantidade do item no carrinho do estoque
            print(f"Estoque temporário após considerar o carrinho:")
            for cafe, qtd in estoque_temporario.items():
                print(f"{cafe}: {qtd} unid. restantes")

            resposta = input("Deseja fechar a conta? (s/n): ").strip().lower()

            if resposta == "s":

                estoque = estoque_temporario  # Atualiza o estoque real com o estoque temporário após a compra
                print(f"Estoque atualizado:")
                for cafe, qtd in estoque.items():
                    print(f"{cafe}: {qtd} unid. restantes")



#               for item in set(carrinho):
#                  if item in estoque:
#                      estoque[item] -= carrinho.count(item)  # Subtrai a quantidade do item no carrinho do estoque
#                 print(f"Estoque atualizado:")
#                 for cafe, qtd in estoque.items():
#                     print(f"{cafe}: {qtd} unid. restantes")

                print(
                    f"\n💸 Conta fechada no valor de R$ {total_conta:.2f}. Obrigado por comprar!")
                break
            else:
                print("🔄 Voltando ao menu principal...")

    elif opcao == "3":
        print("Pedido cancelado. Volte sempre!")
        break
    else:
        print("❌ Opção inválida!")
````

```python
.set()

O que faz: Pega uma lista cheia de itens repetidos e filtra, deixando apenas os itens únicos.

Como ajudou no Totem: Evitou que o seu carrinho mostrasse "Latte" três vezes em linhas separadas. Ele junta e diz apenas "Latte", preparando o terreno para o .count().

```python
.count()
```

O que faz: Conta quantas vezes um elemento específico aparece dentro de uma lista.

Como ajudou no Totem: Permitiu que você descobrisse exatamente quantos cafés de cada tipo o usuário colocou no carrinho (ex: carrinho.count("Latte") te deu o número exato de Lattes).

```python
.copy()
```   

O que faz: Cria uma cópia independente de um dicionário ou lista.

Como ajudou no Totem: Criou um "estoque dublê" (temporário) para você fazer simulações de compra sem estragar o estoque real antes da hora do pagamento.

````python
#v0.03
# Cafeteira_logica

import estoque.itens_estoque as itens_estoque # Importando o seu módulo de estoque

carrinho = []
total_conta = 0.0

print("----- Bem-Vindo à Cafeteria Inteligente! -----")

while True:
    print("""
      =======================
        MENU PRINCIPAL      
      =======================
      
    1. Ver Menu de Cafés
    2. Ver Carrinho e Finalizar Compra
    3. Sair
    4. Ver Estoque (Admin) """)
    
    escolha = input("\nDigite o número da opção desejada: ").strip()

    if escolha == "1":
        # Criamos um loop interno exclusivo para o menu de cafés. 
        # Assim, o 'break' aqui dentro apenas VOLTA para o menu principal!
        while True:
            print("\n------ CAFÉS DISPONÍVEIS ------")
            # Exibe os cafés dinamicamente com base no estoque real e no que já está no carrinho
            for cafe, qtd_total in itens_estoque.estoque.items():
                qtd_disponivel = qtd_total - carrinho.count(cafe)
                print(f"- {cafe}: R$ {itens_estoque.precos[cafe]:.2f} ({qtd_disponivel} unid. disponíveis)")
            
            print("\n[1] Cappuccino | [2] Latte | [3] Espresso | [4] Mocha | [V] Voltar ao Menu Principal")
            opcao = input("\nEscolha o café desejado ou digite 'V': ").strip().lower()
            if opcao == "v":
                print("Saiu com sucesso do menu")
                break

            # Mapeamento da escolha do usuário para o nome do café
            cafes_mapeados = {"1": "Cappuccino", "2": "Latte", "3": "Espresso", "4": "Mocha"}
            
            if opcao in cafes_mapeados:
                nome_cafe = cafes_mapeados[opcao]
                
                # Validação usando sua lógica de estoque temporário/disponível
                if (itens_estoque.estoque[nome_cafe] - carrinho.count(nome_cafe)) > 0:
                    carrinho.append(nome_cafe)
                    total_conta += itens_estoque.precos[nome_cafe]
                    print(f"✅ {nome_cafe} adicionado ao carrinho!")
                else:
                    print(f"❌ Desculpe, o estoque de {nome_cafe} esgotou para este pedido!")
                
            else:
                print("❌ Opção inválida. Escolha de 1 a 4, ou 'v' para voltar.")

    elif escolha == "2":
        print("\n---- SEU CARRINHO ----")
        if not carrinho:
            print("Seu carrinho está vazio!")
        else:
            # Lista os itens sem repetir linhas, mostrando a quantidade
            for cafe in set(carrinho):
                print(f"• {cafe}: {carrinho.count(cafe)} unid. (R$ {itens_estoque.precos[cafe] * carrinho.count(cafe):.2f})")
            print(f"----------------------")
            print(f"Total a Pagar: R$ {total_conta:.2f}")

            opcao_finalizar = input("\nDigite 'Sim' para finalizar ou 'Não' para voltar: ").strip().lower()

            if opcao_finalizar in ["sim", "s"]:
                # MOMENTO 2: Agora sim, damos a baixa definitiva no estoque real!
                for item in set(carrinho):
                    itens_estoque.estoque[item] -= carrinho.count(item)
                
                print("🎉 Compra finalizada com sucesso! Seu café começará a ser preparado.")
                carrinho.clear()
                total_conta = 0.0
            else:
                print("🔄 Voltando ao menu principal. Seus itens continuam salvos no carrinho.")

    elif escolha == "3": 
        print("Obrigado por visitar a Cafeteria Inteligente! Volte sempre! 👋")
        break

    elif escolha == "4":
        print("\n🔒 Acesso restrito ao administrador.")
        admin = input("Digite a senha de administrador: ").strip()
        
        if admin == "admin123":
            print("\n----- ESTOQUE ATUAL REAL -----")
            for cafe, quantidade in itens_estoque.estoque.items():
                print(f"• {cafe}: {quantidade} unidades")
        else:
            print("❌ Senha incorreta. Acesso negado.")
    else:
        print("❌ Opção inválida no menu principal. Escolha de 1 a 4.")
````
